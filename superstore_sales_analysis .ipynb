```python
Q1=df["Sales"].quantile(0.25)
Q3=df["Sales"].quantile(0.75)
IQR=Q3-Q1
print(Q1)
print(Q3)
print(IQR)
lower_bound = Q1 - 1.5 * IQR  # Outlier  اي رقم اقل من كده 
upper_bound = Q3 + 1.5 * IQR  # Outlier  اي رقم اعلى من كده 
print(lower_bound)
print(upper_bound)
```

```python
Sales_outliers = df[df["Sales"] > upper_bound]  #  الصفوف بس الي فوق الحد
Sales_outliers["Sales"].describe()
```

-----------------------------------   OOP ----------------------------------

```python
class SuperAnalysis:
    def __init__(self,df):
        self.df=df
    def category_sales(self):
            return self.df.groupby("Category")["Sales"].sum()
    def category_profit(self):
        return self.df.groupby("Category")["Profit"].sum()
    def yearly_sales(self):
        return self.df.groupby("Year")["Sales"].sum()
    def yearly_profit (self):
        return self.df.groupby("Year")["Profit"].sum()
    def subcategory_sales(self):
        return self.df.groupby("Sub-Category")["Sales"].sum()
    def subcategory_profit(self):
        return self.df.groupby("Sub-Category")["Profit"].sum()
    def monthly_sales(self):
        return self.df.groupby("Month")["Sales"].sum()
    def monthly_profit(self):
        return self.df.groupby("Month")["Profit"].sum()
    def discount_profit(self):
        return self.df.groupby("Discount")["Profit"].mean() 
    def total_sales(self):
        return self.df["Sales"].sum()
    def total_profit(self):
        return self.df["Profit"].sum()
    def total_quantity(self):
        return self.df["Quantity"].sum()
    def total_orders(self):
        return self.df["Order ID"].nunique()
    def avg_discount(self):
        return self.df["Discount"].mean()
    def category_profit_Margin(self):
        return self.df.groupby("Category")["profit Margin"].mean()
    def shipping_profit(self):
        return self.df.groupby("Shipping Days")["Profit"].mean()
    def category_sales_for_item(self):                 
        return self.df.groupby("Category")["Sales for item"].mean()
    def top_category(self):
        return self.category_sales().idxmax()  # اسم الكاتيجوري صاحب اعلى مبيعات
# ----------------- Kpi----------------------
    def Kpi_summary(self):
        return{ #                                    Dict علشان هترجع 
        "Total Sales":self.total_sales(),
        "Total Profit": self.total_profit(),
        "Profit Margin %":(self.total_profit() / self.total_sales()) * 100,
        "Total Orders":self.total_orders(),
        "Avg Discount":self.avg_discount(),
        "Top Category": self.top_category()
    }
    
```
